```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
# Menampilkan data
df = pd.read_excel('Kelulusan Train.xlsx')
df.head(10)
```

```python
# Atribut regular
x = df.drop(['STATUS KELULUSAN'], axis=1)
x.head(10)
```

```python
# Atribut target
y = df['STATUS KELULUSAN']
y.head(10)
```

```python
# Import Naive Bayes
from sklearn.naive_bayes import GaussianNB
```

```python
# Memanggil fungsi Gaussian Naive Bayes
nbc = GaussianNB()
```

```python
df.shape
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

## Filling Null Values

```python
df2 = df.fillna(value = 0)
df2
```

```python
df2.isnull().sum().sum()
```

```python
# Atribut regular
x = df2.drop(['STATUS KELULUSAN'], axis=1)
x.head(10)
```

```python
# Atribut target
y = df2['STATUS KELULUSAN']
y.head(10)
```

```python
# Memasukkan data training pada fungsi klasifikasi Naive Bayes
data_training = nbc.fit(x,y)
```

```python
# Prediksi pada data training
y_predict = data_training.predict(x)
print(y_predict)
```

```python
# Prediksi data testing yang akan diprediksi
JENIS_KELAMIN = 1 #pria
STATUS_MAHASISWA = 0 #mahasiswa
UMUR = 25
STATUS_NIKAH = 1 #menikah
IPS1 = 3.12
IPS2 = 2.75
IPS3 = 2.05
IPS4 = 3.01
IPS5 = 2.56
IPS6 = 3.67
IPS7 = 2.75
IPS8 = 3.45
IPK = np.average ([IPS1, IPS2, IPS3, IPS4, IPS5, IPS6, IPS7, IPS8])
data_testing = [[JENIS_KELAMIN, STATUS_MAHASISWA, UMUR, STATUS_NIKAH, IPS1, IPS2, IPS3, IPS4, IPS5, IPS6, IPS7, IPS8, IPK]]
```

```python
# Prediksi
y_pred = data_training.predict(data_testing)
```

```python
if y_pred == 0:
    hasil = 'Terlambat'
elif y_pred == 1:
    hasil = 'Tepat'
else:
    hasil = 'Error'
print('Hasil Prediksi Kelulusan Mahasiswa =', hasil)
```

```python
# Akurasi model
from sklearn.metrics import accuracy_score
print('Nilai Akurasi = %0.2f' % accuracy_score(y, y_predict))
```

```python
# Hasil
from sklearn.metrics import classification_report
print(classification_report(y, y_predict))
```
